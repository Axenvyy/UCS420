### Q2

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent_values_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    maximum_count = counts.max()
    most_frequent = values[counts == maximum_count]
    indices = {int(value): np.flatnonzero(array == value).tolist() for value in most_frequent}
    return most_frequent, int(maximum_count), indices

for name, array in [("x", x), ("y", y)]:
    frequent_values, frequency, indices = most_frequent_values_and_indices(array)
    print(f"{name}: most frequent value(s) = {frequent_values.tolist()}, count = {frequency}, indices = {indices}")
```

### Q4

```python
student_array = np.linspace(10, 100, 25)
print("Array:", student_array)
print("Dimensions:", student_array.ndim)
print("Shape:", student_array.shape)
print("Total elements:", student_array.size)
print("Element data type:", student_array.dtype)
print("Bytes consumed:", student_array.nbytes)

column_vector = student_array.reshape(25, 1)
print("Transpose using reshape(25, 1):\n", column_vector)
print("Transpose using .T on a 1-D array (unchanged shape):", student_array.T.shape)
print("Transpose of a 2-D row vector using .T:\n", student_array.reshape(1, 25).T)
```

### Q5

```python
ucs420_gurshan = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Array:\n", ucs420_gurshan)
print("Mean:", np.mean(ucs420_gurshan))
print("Median:", np.median(ucs420_gurshan))
print("Maximum:", np.max(ucs420_gurshan))
print("Minimum:", np.min(ucs420_gurshan))
print("Unique elements:", np.unique(ucs420_gurshan))

reshaped_ucs420_gurshan = ucs420_gurshan.reshape(4, 3)
resized_ucs420_gurshan = np.resize(ucs420_gurshan, (2, 3))
print("Reshaped to 4 x 3:\n", reshaped_ucs420_gurshan)
print("Resized to 2 x 3 (first six values):\n", resized_ucs420_gurshan)
```
