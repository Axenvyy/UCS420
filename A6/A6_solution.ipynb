### Q1

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_celsius = temperature + 2
fahrenheit = corrected_celsius * 9 / 5 + 32
above_32 = corrected_celsius[corrected_celsius > 32]
count_above_32 = np.count_nonzero(corrected_celsius > 32)

print("Corrected Celsius readings:", corrected_celsius)
print("Fahrenheit readings:", fahrenheit)
print("Corrected readings above 32 C:", above_32)
print("Count above 32 C:", count_above_32)
print("Vectorized operations apply calculations to the whole array without Python-level iteration. Boolean indexing selects matching readings directly, making the code concise and efficient for large arrays.")
```

### Q2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

maximum_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total steps per day:", np.sum(steps, axis=0))
print("Total steps per user:", np.sum(steps, axis=1))
print("Maximum position (zero-based user, day):", maximum_position)
```

### Q3

```python
original = np.array([1, 2, 3, 4, 5, 6])
subset = original[1:4]
subset[0] = 999
print("Original after modifying view:", original)
print("Subset view:", subset)

copied_subset = original[1:4].copy()
copied_subset[0] = 500
print("Original after modifying copy:", original)
print("Copied subset:", copied_subset)

matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:\n", matrix)
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2, columns 2-3 (one-based):\n", matrix[0:2, 1:3])

flat_array = matrix.flatten()
flat_array[0] = 500
print("Flattened copy after modification:", flat_array)
print("Matrix unchanged by flatten():\n", matrix)
ravel_array = matrix.ravel()
ravel_array[1] = 999
print("Raveled view after modification:", ravel_array)
print("Matrix changed by ravel():\n", matrix)
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

### Q4

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)
y = np.array([40, 65, 30, 85], dtype=float)

X_transpose = X.T
xtx = X_transpose @ X
xtx_inverse = np.linalg.inv(xtx)
coefficients = xtx_inverse @ X_transpose @ y
new_user = np.array([5, 40, 7], dtype=float)
predicted_score = new_user @ coefficients

print("X shape:", X.shape)
print("X dimensions:", X.ndim)
print("X transpose (rows are features, columns are users):\n", X_transpose)
print("X.T @ X:\n", xtx)
print("Inverse of X.T @ X:\n", xtx_inverse)
print("OLS coefficients [sleep, activity, stress]:", coefficients)
print("Each coefficient is the model's score change per one-unit increase in that feature, holding the other features fixed. This fit has no intercept, as in the assigned equation.")
print("Predicted assistance score:", predicted_score)
```
